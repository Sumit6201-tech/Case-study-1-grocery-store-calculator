```python
def calculate_total_cost():

    total_cost = 0

    num_items = int(input("How many items are you purchasing? "))

    for i in range(0,num_items,1):
        print(f"\nItem {i+1}:")
        item_name = input("Enter item name: ")
        price_per_item = float(input(f"Enter price per {item_name}: "))
        quantity = int(input(f"Enter quantity for {item_name}: "))

        item_total = price_per_item * quantity
        total_cost += item_total

        print(f"Total cost for {item_name}: ${item_total:.2f}")

    print(f"\nTotal Bill: ${total_cost:.2f}")

calculate_total_cost()
```
